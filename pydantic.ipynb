```python
from dataclasses import dataclass, field
@dataclass
class Checking_data:
    name: str 
    age: int = field(default=25)
    email: str = field(default="example@example.com")
```

```python
xyz = Checking_data()
print(xyz)
```

```python
sandy_data = Checking_data(name="John Doe",
              age=30,
              email="sandeep@example.com")
print(sandy_data)
print(sandy_data.name)
print(sandy_data.age)
print(sandy_data.email)
```

```python
meenu_data = Checking_data(name=35, age='25', email="meenu@gmail.com")
print(meenu_data)
print(meenu_data.name)
print(meenu_data.age)
print(meenu_data.email)
```

Disadvantages of DataClass
1) Type checking is not happening
2) also, you can only define the default value to the field only after non default value. i.e first you will have to define non defualt field, then only you can define default value field.

To overcome the issue of type checking, pydantic is being used.

Below is the code for Pydantic data validation

```python
from pydantic import BaseModel, Field
class Checking_data_pydantic(BaseModel):
    name: str
    age: int = Field(default=25)
    email: str
    
```

```python
abc = Checking_data_pydantic(name='Dhairya', age='28', email='10')
print(abc)
```

```python
from typing import Optional
from pydantic import BaseModel
class pydantic_checking(BaseModel):
    name: str = Field(default="Dhairya")
    age: Optional[int]
    email: Optional[str] = "Noneacb@gmail.co"
  
```

```python
abc = pydantic_checking(age=35)
print(abc)
```

Difference bewteen Optional and Field:

Use Optional to indicate a field may be missing or None or set any default value
Use Field to set constraints, defaults, and metadata, gt, lt, ge etc.
They serve different purposes but can be used together.

Advantages of Pydantic:

1) type checking is happening on the fly
2) Type casting is also taken care of, whereever possible e.g '35' can be turned into 35 if the field is of integet type.
3) Pydantic is very fast because it is written in rust framework 

# Model with nested Model
Create Complex structure with Nested Model

```python
from pydantic import BaseModel, Field
class nested_model_marks(BaseModel):
    english: int = Field(default=None, lt=100)
    hindi: int = Field(default=None, lt=50)
    science: int = Field(default=None, lt=100)
    maths: int = Field(default=None, gt=50)
    social_science: int = Field(default=None, lt=100)
```

```python
class student_report_card(BaseModel):
    name: str
    roll_number: int
    marks: nested_model_marks
```

```python
student1 = student_report_card(
    name="Dhairya",
    roll_number=101,
    marks=nested_model_marks(
        english=85,
        hindi=45,
        science=90,
        maths=60,
        social_science=80
    )
)
print(student1)
```
