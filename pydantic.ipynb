```python
from dataclasses import dataclass, field
@dataclass
class Checking_data:
    name: str 
    age: int = field(default=25)
    email: str = field(default="example@example.com")
```

```python
xyz = Checking_data()
print(xyz)
```

```python
sandy_data = Checking_data(name="John Doe",
              age=30,
              email="sandeep@example.com")
print(sandy_data)
print(sandy_data.name)
print(sandy_data.age)
print(sandy_data.email)
```

```python
meenu_data = Checking_data(name=35, age='25', email="meenu@gmail.com")
print(meenu_data)
print(meenu_data.name)
print(meenu_data.age)
print(meenu_data.email)
```
